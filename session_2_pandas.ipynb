```python
visa_df  #df
visa_df['continent']  #one column
con=visa_df['continent']=='Asia'  #one class
np.sum(con)  #adding
visa_df[con] #extraction
```

```python
freq=[]
for i in visa_df['continent'].unique():
    con=visa_df['continent']==i
    freq.append(np.sum(con))
df=pd.DataFrame(zip(visa_df['continent'].unique(),freq),columns=['class','frequency'])
df
```

**value_counts**

```python
visa_df['continent'].value_counts()
```

```python
plt.bar(df['class'],df['frequency'])
plt.xlabel('class')
plt.ylabel('frequency')
plt.title('Applications from each continent')
plt.show()
```

```python
keys=visa_df['continent'].value_counts().keys()
values=visa_df['continent'].value_counts().values
pd.DataFrame(zip(keys,values),columns=['class','frequency'])
```
