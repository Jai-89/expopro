**Emperical Rule**

```python
- 68% of data available between u-1*sd to u+1*sd
- step-1: calculate mean of wage
- step-2: calculate sd of wage
- step-3: lb=mean-1*sd
- step-4: ub=mean+1*sd
- step-5: con1=wage_data>lb
- step-6: con2= wage_data<ub
- step-7: con= con1&con2
- step-8: np.sum(con)
- step-9: 68%(25480)
```

```python
Mean=wage_data.mean()
Mean
sd=wage_data.std()
mean
```

```python
lb=Mean-1*sd
ub=Mean+1*sd
con1=wage_data>lb
con2= wage_data<ub
con= con1&con2
np.sum(con)==68*(25480)/100
```
